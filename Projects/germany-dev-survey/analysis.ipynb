## 1. How many developers from Germany answered?

```python
de = df[df["Country"] == "Germany"]
print(len(de), "developers from Germany")
print(round(len(de) / len(df) * 100, 1), "% of all answers")
```

About 6.1% of all answers come from Germany, so we have 3,025 German developers to analyse.

```python
langs = de["LanguageHaveWorkedWith"].str.split(";").explode()
top10 = langs.value_counts().head(10)
top10
```

```python
top10.sort_values().plot(kind="barh")
plt.title("Top 10 languages used by developers in Germany")
plt.xlabel("Number of developers")
plt.show()
```

Python is the 3rd most used language among German developers. The top 3 are JavaScript, HTML/CSS and Python.
